# Dynamic Social Networks

## 1. Introduction

Traditional networks are static, but real-world networks evolve over time (e.g. social ties form/break). Dynamic Changes can happen such as users join or leave, interaction strength fluctuates. These changes affect processes like information or disease spread, centrality, and clustering.

* **Use Cases & Examples**:

  * Social media interactions and friendship evolution.
  * Wildlife tracking and epidemic simulations (e.g., crocodiles, herons).
  * Human systems under stress (e.g., COVID-19, Ukraine conflict).
    <br></br>

* **Social Media Impact**:

  * Hashtags, tags, and groups expand users' networks.
  * Algorithms can unintentionally promote: Misinformation, Racist or harmful ideologies,Feedback loops of unhealthy content.
      <br></br>

* **Triadic Closure**:

  * A key mechanism in network evolution.
  * Based on the idea that **“friends of your friends are likely to become your friends.”**
  * Two forms: **Connection-based** where users are likely to connect if they share mutual friends. **Metadata-based** where users with shared interests are suggested to each other (e.g., football fans, hobby groups).
  * High **open triads** in a network signal strong potential for growth and new connections.

## 4. Dynamic SIR model 

This section expands on Chapter 3 by introducing the mathematical foundation of SIR models, using differential equations to describe how disease spreads and recovers over time based on transmission and recovery rates. It also notes the limitations of assuming random interactions and highlights the value of dynamic networks for more realistic epidemic modeling.

$\displaystyle\frac{dS}{dt}=-\frac{\beta I S}{N}$<br>

$\displaystyle\frac{dI}{dt}=\frac{\beta I S}{N} - \gamma I $<br>

$\displaystyle\frac{dR}{dt}=\gamma I$<br>

Where: 
* $\beta$ is the transmission rate
* $\gamma$ is the recovery rate
* S: Susceptibal individual number
* I: Infenfected individual number
* R: Recovered individual who can get infected again number
* N = S+I+R ; The total popualtion


Using forward Euler integration, update equations become:


$\displaystyle S_{n+1} = S_n - \Delta t \cdot \beta \frac{S_n I_n}{N}$ <br>

$\displaystyle I_{n+1} = I_n + \Delta t \cdot \left( \beta \frac{S_n I_n}{N} - \gamma I_n \right)$  <br>

$\displaystyle R_{n+1} = R_n + \Delta t \cdot \gamma I_n$ <br>




### Simple version without graphs

```python
def sir_model(y, t, beta, gamma):
    S, I, R = y
    N = S + I + R
    dSdt = -beta * S * I / N
    dIdt = beta * S * I / N - gamma * I
    dRdt = gamma * I
    return [dSdt, dIdt, dRdt]

from scipy.integrate import odeint
import numpy as np

# Initial values
S0, I0, R0 = 990, 10, 0
beta, gamma = 0.3, 0.15
t = np.linspace(0, 160, 160)  # Time grid (days)

# Solve the ODE
solution = odeint(sir_model, [S0, I0, R0], t, args=(beta, gamma))
```

```python
# Assuming beta and gamma are defined
R0 = beta / gamma

plt.plot(solution[:, 0], label='Susceptible', linestyle='--')
plt.plot(solution[:, 1], label='Infected', marker='*', linestyle='-')
plt.plot(solution[:, 2], label='Recovered', linestyle='dotted')

plt.xlabel("Time (days)")
plt.ylabel("Population count")
plt.title(f"SIR Model (R₀ = {R0:.2f})")
plt.legend()
plt.grid(True)
plt.show()
```

## 5. Factors influencing the spread 

#### 🔑 Key points to remember


1. **Centrality Metrics for Epidemic Modeling:**
    - **Degree centrality**: Measures a node’s connectivity—key for modeling local spread.
    - **PageRank & hubness centrality**: Highlight influential spreaders in the network.
    - **Betweenness centrality**: Identifies bridge nodes that connect communities—critical for modeling cross-community transmission.
<br><br>


2. **Structural Roles in Networks:**
   * **Hubs** (high-degree nodes) rapidly amplify spread within communities.
   * **Bridges** (high-betweenness nodes) enable spread between communities or regions.
<br><br>

3. **Advanced Metric – Forman-Ricci Curvature:**
   * Integrates both hub and bridge properties to identify key nodes/edges for intervention.
   * Useful in simulations and real-world network-based epidemic control strategies.
<br><br>

4. **Global Network Measures:**
   * **Eccentricity**: is the maximum distance from a vertex to another vertex in a graph, is another metric commonly used to measure the potential impact of an epidemic at a population level.
   * **Radius**: The smallest eccentricity value across network vertices
   * **Diameter**: The largest eccentricity value across network vertices.
   * **Dense networks** have smaller diameters than **sparse networks**, as vertices tend to be more interconnected (allowing for greater spread potential inshorter periods of time).

Example we have a graph (G): 
- path(A): node 1 -> node 2 =>  4 node jumps
- path(B): node 1 -> node 2 =>  10 node jumps 
- path(C): node 3 -> node 4 =>  5 node jumps 
- path(D): node 3 -> node 4 =>  2 node jumps 

$Eccentricity(1,2) = max(path(A),path(B)) = path(B)=10$<br>
$Eccentricity(3,4) = max(path(C),path(D)) = path(C)=5$<br>
$Radius(G) = min(all\,paths)=path(D)=2$<br>
$Diameter(G) = max(all\,paths)=path(B)=10$<br>

### Run SIR model large sample

```python
#run epidemic model
import ndlib.models.ModelConfig as mc
%matplotlib inline
from ndlib.viz.mpl.DiffusionTrend import DiffusionTrend
import ndlib.models.dynamic as dm
```

```python
# model selection
model = dm.DynSIRModel(crocodile)

# model configuration
config = mc.Configuration()
config.add_model_parameter('beta', 0.3) # infection rate
config.add_model_parameter('gamma', 0.05) # recovery rate
config.add_model_parameter("percentage_infected", 0.25)
model.set_initial_status(config)

# simulate snapshot based execution
iterations = model.execute_snapshots()
#iterations = model.execute_iterations()
trends = model.build_trends(iterations)

#visualize trends
viz = DiffusionTrend(model, trends)
viz.plot()
```

As time went by more connections were instablished which cause the infection rates to increase. 

```python
#obtain network centrality statistics for crocodile network
data = []
nodes = []
max_degree = [] 
diameter = []
radius = [] 

for i in range(10):

    # iterate to fill up metadata
    dg = crocodile.time_slice(i)

    # fill ip distribution values for betweeness
    try:
        dd = nx.betweenness_centrality(dg)
        _ = pd.DataFrame(list(dd.values()),columns=['Betweeness Centrality'])
        _['time'] = f"T={i}"
        data.append(_)
    except:
        # fill ip distribution values for betweeness
        data.append([])
        

    try: nodes.append(len(dg.nodes()))
    except: nodes.append(np.nan)
        
    try: max_degree.append(max(dg.degree().values()))
    except: max_degree.append(np.nan)
        
    try:diameter.append(nx.diameter(dg))
    except: diameter.append(np.nan)      
    
    try: radius.append(nx.radius(dg))
    except: radius.append(np.nan)


res = pd.DataFrame(data={'nodes':nodes,
                         'max_degree':max_degree,
                         'diameter':diameter,
                         'radius':radius})
df = pd.concat(data)
```

```python
res.plot(subplots=True, layout=(2, 2), figsize=(10, 6),sharex=False)
```

## 8. Extra -> Run SIR model for large network using Barabási–Albert Growth

```python
# -------------------------------
# 1. Reproducibility
random_seed = 42
random.seed(random_seed)
np.random.seed(random_seed)

# -------------------------------
# 2. Parameters
num_nodes = 100
timesteps = 20
batch_size = 5   # new nodes added per timestep
m = 2             # edges per new node in BA model

# -------------------------------
# 3. Create dynamic graph with progressive BA growth
dynG = dn.DynGraph()
all_nodes = list(range(num_nodes))
added_nodes = set()

current_node_index = 0

for t in range(timesteps):
    if current_node_index + batch_size > num_nodes:
        break

    # Add batch of new nodes
    new_nodes = list(range(current_node_index, current_node_index + batch_size))
    added_nodes.update(new_nodes)

    # Generate BA graph with all current nodes
    G = nx.barabasi_albert_graph(len(added_nodes), m, seed=random_seed + t)

    # Map BA graph node indices to actual node IDs in dynG
    node_map = dict(zip(G.nodes(), sorted(added_nodes)))
    for u, v in G.edges():
        dynG.add_interaction(node_map[u], node_map[v], t)

    current_node_index += batch_size
```

```python
# Create side-by-side subplots
fig, axes = plt.subplots(3, 3, figsize=(30, 30))
axes = axes.flatten()

# Create plots for each time point
for i in range(9):
    g1 = dynG.time_slice(i)
    nx.draw(g1, ax=axes[i], with_labels=True,node_size=500,font_size=30)
    axes[i].set_title(f"t={i}",fontsize=30)

plt.tight_layout()
plt.subplots_adjust(wspace=0.4, hspace=0.4)  # Add horizontal and vertical spacing
plt.show()
```

```python
# model selection
model = dm.DynSIRModel(dynG)

# model configuration
config = mc.Configuration()
config.add_model_parameter('beta', 0.3) # infection rate
config.add_model_parameter('gamma', 0.05) # recovery rate
config.add_model_parameter("percentage_infected", 0.25)
model.set_initial_status(config)

# simulate snapshot based execution
iterations = model.execute_snapshots()
#iterations = model.execute_iterations()
trends = model.build_trends(iterations)

#visualize trends
viz = DiffusionTrend(model, trends)
viz.plot()
```

As the newtwork becomes more dense and connected, the infections rates increase and so does the recovered. 

```python
#obtain network centrality statistics for crocodile network
data = []
nodes = []
max_degree = [] 
diameter = []
radius = [] 

for i in range(12):
    try:
        # iterate to fill up metadata
        dg = dynG.time_slice(i)

        # fill ip distribution values for betweeness
        dd = nx.betweenness_centrality(dg)
        _ = pd.DataFrame(list(dd.values()),columns=['Betweeness Centrality'])
        _['time'] = f"T={i}"
        data.append(_)
        

        # append metadata
        nodes.append(len(dg.nodes()))
        max_degree.append(max(dg.degree().values()))
        diameter.append(nx.diameter(dg))
        radius.append(nx.radius(dg))
    except:
        # fill ip distribution values for betweeness
        data.append([])

        # append metadata
        nodes.append(np.nan)
        max_degree.append(np.nan)
        diameter.append(np.nan)
        radius.append(np.nan)

res = pd.DataFrame(data={'nodes':nodes,
                         'max_degree':max_degree,
                         'diameter':diameter,
                         'radius':radius})
df = pd.concat(data)
```
